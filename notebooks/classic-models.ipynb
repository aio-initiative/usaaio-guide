## Your task

With **NumPy only** (no `sklearn.linear_model`):

1. Initialize `w = 0`, `b = 0`. For a fixed number of epochs, loop over training points. On each misclassified point (`y * (w @ x + b) <= 0`) update `w += y * x`, `b += y`.
2. Set `predictions` to `sign(X_test @ w + b)` as ±1 labels (map 0 to +1).

## Evaluation

Test accuracy must reach the target shown in the evaluation cell.


````python
import numpy as np

rng = np.random.default_rng(191)
d = 10
w_true = rng.normal(0, 1, d); w_true /= np.linalg.norm(w_true)

def make(n, seed):
    r = np.random.default_rng(seed)
    X = r.normal(size=(n, d))
    margin = X @ w_true
    keep = np.abs(margin) > 0.25          # carve out a margin
    X, margin = X[keep], margin[keep]
    y = np.sign(margin).astype(int)
    flip = r.random(len(y)) < 0.02
    return X, np.where(flip, -y, y)

X_train, y_train = make(1900, 201)
X_train, y_train = X_train[:1500], y_train[:1500]
X_test, y_test = make(650, 202)
X_test, y_test = X_test[:500], y_test[:500]
print(f"X_train {X_train.shape}, labels {np.unique(y_train)}")
````

## Solution

The classic perceptron rule: sweep the data, and on every misclassified point add `y * x` to the weights. A handful of epochs is plenty. More just chases the 2% flipped labels.

```python
import numpy as np

w = np.zeros(X_train.shape[1])
b = 0.0
for _epoch in range(10):
    for x, y in zip(X_train, y_train):
        if y * (w @ x + b) <= 0:
            w += y * x
            b += y

raw = np.sign(X_test @ w + b)
predictions = np.where(raw == 0, 1, raw).astype(int)
```
