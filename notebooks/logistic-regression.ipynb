## Solution

Append a bias column, start weights at zero, and run batch gradient descent on binary cross-entropy, whose gradient is the beautifully simple `Xᵀ(σ(Xw) - y)/n`. Threshold the test sigmoid at 0.5.

```python
import numpy as np

def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

Xb = np.column_stack([X_train, np.ones(len(X_train))])
w = np.zeros(Xb.shape[1])
for _ in range(4000):
    p = sigmoid(Xb @ w)
    w -= 0.1 * Xb.T @ (p - y_train) / len(y_train)

Xt = np.column_stack([X_test, np.ones(len(X_test))])
predictions = (sigmoid(Xt @ w) >= 0.5).astype(int)
```
