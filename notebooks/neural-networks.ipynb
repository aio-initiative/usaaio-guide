````python
import numpy as np
import matplotlib.pyplot as plt

def spirals(n, seed):
    r = np.random.default_rng(seed)
    n2 = n // 2
    t = r.uniform(0.4, 3.0 * np.pi, n2)
    x0 = np.column_stack([t * np.cos(t), t * np.sin(t)]) / 3.0
    x1 = np.column_stack([t * np.cos(t + np.pi), t * np.sin(t + np.pi)]) / 3.0
    X = np.vstack([x0, x1]) + r.normal(0, 0.12, (n2 * 2, 2))
    y = np.array([0] * n2 + [1] * n2)
    idx = r.permutation(len(y))
    return X[idx], y[idx]

X_train, y_train = spirals(2000, 111)
X_test, y_test = spirals(600, 112)
print(f"X_train {X_train.shape}: try plt.scatter(X_train[:,0], X_train[:,1], c=y_train, s=4)")
````

## Solution

Spirals are not linearly separable, but a small two-hidden-layer MLP bends the decision boundary around them easily. Standardize inputs and give the optimizer enough iterations.

```python
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler().fit(X_train)
mlp = MLPClassifier(hidden_layer_sizes=(64, 64), activation="tanh",
                    max_iter=3000, random_state=0)
mlp.fit(scaler.transform(X_train), y_train)
predictions = mlp.predict(scaler.transform(X_test))
```
