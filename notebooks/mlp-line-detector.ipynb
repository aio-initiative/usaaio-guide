# Build a line detector by hand

A neuron that reads a $3 \times 3$ image has nine weights, one per pixel, and a bias. It
multiplies each pixel by its weight, adds everything up with the bias, and passes the total
through a sigmoid. Here you pick the ten numbers yourself, with no training, so that the
neuron fires on the image whose center column is lit and stays quiet on five other images.

**How to work through it.** Each part asks for one function. Replace the `TODO` and the
`raise NotImplementedError` with your code, then run the check cell under it. The check
prints `Part N passed` when your function is right, and stops with a message that says
what is wrong when it is not.

## Setup

Run this cell first.

```python
import numpy as np

images = {
    "center column": [[0, 1, 0], [0, 1, 0], [0, 1, 0]],
    "nothing":       [[0, 0, 0], [0, 0, 0], [0, 0, 0]],
    "left column":   [[1, 0, 0], [1, 0, 0], [1, 0, 0]],
    "middle row":    [[0, 0, 0], [1, 1, 1], [0, 0, 0]],
    "everything":    [[1, 1, 1], [1, 1, 1], [1, 1, 1]],
    "center pixel":  [[0, 0, 0], [0, 1, 0], [0, 0, 0]],
}
images = {name: np.array(pixels, dtype=float) for name, pixels in images.items()}


def neuron(image, W, b):
    """Sigmoid of the weighted pixel total plus the bias."""
    total = np.sum(W * image) + b
    return 1 / (1 + np.exp(-total))


for name in images:
    rows = ["".join("#" if p else "." for p in row) for row in images[name]]
    print(f"{name:14s}", "  ".join(rows))
```

## Solution

### Part 1: Choose the weights

```python
def line_detector():
    """(W, b): a (3, 3) array of weights and a bias, chosen by hand."""
    W = np.array([[-4.0, 4.0, -4.0],
                  [-4.0, 4.0, -4.0],
                  [-4.0, 4.0, -4.0]])
    b = -8.0
    return W, b
```

**Why it works.** Each pixel of the center column earns 4 when lit, 12 for the three together,
and the bias takes 8 away, so the target image reaches a total of 4 and an output of 0.98.
Every other pixel costs 4 when lit. `"everything"` earns the same 12 from the column but pays
24 for the rest, a total of $-20$. `"center pixel"` earns only 4, less than the 8 the bias
demands, so its total is $-4$ and the output is 0.02. `"middle row"` lights one column pixel
and two others, for $4 - 8 - 8 = -12$. Positive weights along the shape you want and negative
weights everywhere else make a template, and the first hidden layer of a digit classifier is
a stack of templates like this one, each with 784 weights.
