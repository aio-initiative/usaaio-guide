## Solution


### Problem 1

A separating line that passes close to some training points gets them right by a hair, and a small shift, such as the noise in a new sample drawn near them, carries such a point across. The max-margin line keeps the largest possible distance to the nearest point of each class, so every training point can move by less than half the margin width without changing sides, and new points that land near the training points are classified the same way.

The line is also decided by the support vectors alone. Points far from the boundary can move, or be added or removed, without changing it. And a wider margin means a shorter $\mathbf{w}$, so maximizing the margin acts like the weight penalty of ridge regression: among all the lines that fit the training data it picks the simplest one, and simpler models tend to generalize better.


### Problem 2

The soft-margin SVM minimizes $\tfrac12\lVert\mathbf{w}\rVert^2 + C\sum_i \xi_i$ subject to $y_i(\mathbf{w}^\top\mathbf{x}_i + b) \ge 1 - \xi_i$ and $\xi_i \ge 0$. Write $m_i = y_i(\mathbf{w}^\top\mathbf{x}_i + b)$. For fixed $\mathbf{w}$ and $b$ the two constraints say $\xi_i \ge 1 - m_i$ and $\xi_i \ge 0$, that is $\xi_i \ge \max(0,\ 1 - m_i)$. The objective grows with every $\xi_i$, so at the optimum each slack sits at its lower bound, $\xi_i = \max(0,\ 1 - m_i)$. Substituting it removes the constraints:

$$
\min_{\mathbf{w}, b}\ \tfrac12\lVert\mathbf{w}\rVert^2 + C\sum_i \max\big(0,\ 1 - y_i(\mathbf{w}^\top\mathbf{x}_i + b)\big).
$$

The term $\max(0,\ 1 - m)$ is the hinge loss. It is zero for a point with $m \ge 1$, outside the margin on its own side, and rises in a straight line with slope $-1$ as $m$ falls below 1.


### Problem 3

Small $C$ makes slack cheap, so the margin is wide and many points sit inside it: at $C = 0.01$ there are 106 support vectors and the margin is 2.38 wide in the scaled features. As $C$ grows each violation costs more, the margin narrows and the support vectors thin out, down to 4 at $C = 100$ with a margin of 0.40. The two blobs are almost separable, so the training accuracy is 0.996 at every $C$.


```python
import numpy as np
import matplotlib.pyplot as plt

from sklearn.datasets import make_blobs
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.svm import SVC

def plot_boundary(model, X, y, ax, title):
    """Decision regions, the boundary (score 0) and the margin edges (score -1 and +1)."""
    x0 = np.linspace(X[:, 0].min() - 1, X[:, 0].max() + 1, 300)
    x1 = np.linspace(X[:, 1].min() - 1, X[:, 1].max() + 1, 300)
    g0, g1 = np.meshgrid(x0, x1)
    score = model.decision_function(np.c_[g0.ravel(), g1.ravel()]).reshape(g0.shape)
    ax.contourf(g0, g1, score > 0, alpha=0.2, cmap="coolwarm")
    ax.contour(g0, g1, score, levels=[-1, 0, 1], colors="k", linestyles=["--", "-", "--"])
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap="coolwarm", s=12, edgecolors="k", linewidths=0.3)
    ax.set_title(title)

X, y = make_blobs(
    n_samples=250,
    centers=2,
    cluster_std=2,
    random_state=42
)

C_values = [0.01, 0.1, 1, 10, 100]

linear_models = {}

fig, axes = plt.subplots(1, len(C_values), figsize=(4 * len(C_values), 4))
for ax, C in zip(axes, C_values):
    model = Pipeline([("scaler", StandardScaler()), ("svm", SVC(kernel="linear", C=C))])
    model.fit(X, y)
    linear_models[C] = model
    width = 2 / np.linalg.norm(model[-1].coef_)
    print(f"C = {C:>6}: {model[-1].n_support_.sum():3d} support vectors, "
          f"margin {width:.2f} (scaled features), train accuracy {model.score(X, y):.3f}")
    plot_boundary(model, X, y, ax, f"C = {C}")
plt.tight_layout()
plt.show()
```

### Problem 4

With $\gamma = 0.01$ and $0.1$ every point reaches far, the boundary is close to a line and cuts through both moons, and the training accuracy is only 0.86 and 0.87: the model underfits. At $\gamma = 1$ the boundary follows the two moons (0.96). At $\gamma = 100$ each point reaches only its near neighbors, the boundary breaks into small islands around single points, 279 of the 300 points are support vectors and the training accuracy climbs to 0.987 by fitting the noise: the model overfits. Training accuracy alone cannot show the overfitting, since it keeps rising with $\gamma$. Held-out data, as in Problem 5, shows the large values doing worse.


```python
from sklearn.datasets import make_moons

X, y = make_moons(
    n_samples=300,
    noise=0.22,
    random_state=42
)

gamma_values = [0.01, 0.1, 1, 10, 100]

rbf_models = {}

fig, axes = plt.subplots(1, len(gamma_values), figsize=(4 * len(gamma_values), 4))
for ax, gamma in zip(axes, gamma_values):
    model = Pipeline([("scaler", StandardScaler()), ("svm", SVC(kernel="rbf", gamma=gamma))])
    model.fit(X, y)
    rbf_models[gamma] = model
    print(f"gamma = {gamma:>6}: {model[-1].n_support_.sum():3d} support vectors, "
          f"train accuracy {model.score(X, y):.3f}")
    plot_boundary(model, X, y, ax, f"gamma = {gamma}")
plt.tight_layout()
plt.show()
```

### Problem 5

Searching $C \in \{0.1, 1, 10, 100\}$ and $\gamma \in \{0.01, 0.1, 1, 10, 100\}$ with 5-fold cross-validation picks $C = 10$ and $\gamma = 1$, with a mean validation accuracy of 0.929. The tuned model scores 0.973 on the test set, which the search never saw.


```python
from sklearn.model_selection import train_test_split, GridSearchCV

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25,random_state=42)

param_grid = {
    "svm__C": [0.1, 1, 10, 100],
    "svm__gamma": [0.01, 0.1, 1, 10, 100],
}

grid = GridSearchCV(
    Pipeline([("scaler", StandardScaler()), ("svm", SVC(kernel="rbf"))]),
    param_grid,
    cv=5,
    scoring="accuracy"
)

grid.fit(X_train, y_train)

print("best parameters:", grid.best_params_)
print(f"mean validation accuracy: {grid.best_score_:.3f}")
test_acc = grid.score(X_test, y_test)
print(f"test accuracy: {test_acc:.3f}")
```

### Problem 6

Logistic regression and the linear SVM both draw a straight line, and both score 0.920 on the 75 test points. The RBF SVM scores 0.933, one more test point right. One point is a thin difference on its own, but the training accuracies say the same thing more clearly: 0.86 and 0.85 for the two linear models against 0.92 for the RBF SVM, so the straight lines underfit the curved moons. Choose the RBF SVM here, with $C$ and $\gamma$ tuned by cross-validation as in Problem 5. On data that a line separates well, logistic regression would be the better pick, since it is as accurate, faster to train, and gives probabilities.


```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

models = {
    "Logistic Regression": Pipeline([
        ("scaler", StandardScaler()),
        ("model", LogisticRegression())
    ]),
    "Linear SVM": Pipeline([
        ("scaler", StandardScaler()),
        ("model", SVC(kernel="linear"))
    ]),
    "RBF SVM": Pipeline([
        ("scaler", StandardScaler()),
        ("model", SVC(kernel="rbf"))
    ])
}

test_accs = {}

for name, model in models.items():
    model.fit(X_train, y_train)
    test_accs[name] = accuracy_score(y_test, model.predict(X_test))
    print(f"{name:<20} train {model.score(X_train, y_train):.3f}, test {test_accs[name]:.3f}")
```
