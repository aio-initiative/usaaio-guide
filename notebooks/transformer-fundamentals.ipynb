## Solution

Build the angle table `pos / 10000^(2i/d)` with one broadcasted division, then write `sin` into the even columns and `cos` into the odd columns.

```python
import numpy as np

pos = np.arange(SEQ_LEN)[:, None]
i = np.arange(D_MODEL // 2)[None, :]
angles = pos / np.power(10000.0, 2 * i / D_MODEL)

pos_encoding = np.zeros((SEQ_LEN, D_MODEL))
pos_encoding[:, 0::2] = np.sin(angles)
pos_encoding[:, 1::2] = np.cos(angles)
print(pos_encoding.shape)
```
