````python
import numpy as np

rng = np.random.default_rng(5)
n_train, n_test, d = 800, 300, 12

mix = rng.normal(size=(d, d))          # makes the features correlated
w_true = rng.normal(0, 2.0, d)
b_true = 3.5

def make(n, seed):
    r = np.random.default_rng(seed)
    Z = r.normal(size=(n, d))
    X = Z @ mix
    y = X @ w_true + b_true + r.normal(0, 4.0, n)
    return X, y

X_train, y_train = make(n_train, 10)
X_test, y_test = make(n_test, 20)
print(f"X_train {X_train.shape}, X_test {X_test.shape}")
````
