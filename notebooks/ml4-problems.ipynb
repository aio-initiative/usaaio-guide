### Problem 1

### Problem 2

The entropy of the parent is $-(0.7\log_2 0.7 + 0.3\log_2 0.3) = 0.8813$ bits. The left child has shares $5/6$ and $1/6$, entropy $0.6500$ bits, and the right child is an even split, entropy $1$ bit. The children hold $6/10$ and $4/10$ of the examples, so the weighted child entropy is $0.6(0.6500) + 0.4(1) = 0.7900$ bits and the information gain is $0.8813 - 0.7900 = 0.0913$ bits.


```python
import numpy as np

def entropy(counts):
    p = np.array(counts) / sum(counts)
    p = p[p > 0]
    return float(-(p * np.log2(p)).sum())

info_gain = entropy([7, 3]) - 0.6 * entropy([5, 1]) - 0.4 * entropy([2, 2])
print(f"information gain: {info_gain:.4f} bits")
```

### Problem 3

Training accuracy can only rise as the depth grows, and the unlimited tree reaches 1.00. Test accuracy rises at first, then falls back once the extra depth starts to fit the noise in the training rows.


```python
import pandas as pd
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score

X, y = make_moons(n_samples=300,noise=0.25,random_state=42)
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.25,random_state=42)

depth_values = [1, 2, 3, 4, 6, 8, None]

depth_results = []

for depth in depth_values:
    tree = DecisionTreeClassifier(max_depth=depth, random_state=42)
    tree.fit(X_train, y_train)
    depth_results.append({
        "max_depth": depth,
        "train_acc": accuracy_score(y_train, tree.predict(X_train)),
        "test_acc": accuracy_score(y_test, tree.predict(X_test)),
    })

print(pd.DataFrame(depth_results))
```

### Problem 4

Each tree is fit on a bootstrap sample, so about a third of the training rows are left out of it. The OOB score predicts every training row with only the trees that did not see it, which makes it a validation estimate that costs no held-out rows. It should land close to the test accuracy.


```python
from sklearn.ensemble import RandomForestClassifier

forest = RandomForestClassifier(
    n_estimators=200,
    oob_score=True,
    random_state=42
)
forest.fit(X_train, y_train)

forest_test_acc = accuracy_score(y_test, forest.predict(X_test))
forest_oob = forest.oob_score_
print(f"test accuracy {forest_test_acc:.3f}, OOB score {forest_oob:.3f}")
```

### Problem 5

The single tree is grown to full depth, so it fits the training set exactly and gives up accuracy on the test set. The forest averages many such trees and the boosted model adds many shallow ones, and both hold up better on the test set than the single tree.


```python
from sklearn.ensemble import GradientBoostingClassifier

models = {
    "Decision Tree": DecisionTreeClassifier(random_state=42),
    "Random Forest": RandomForestClassifier(n_estimators=200, random_state=42),
    "Gradient Boosting": GradientBoostingClassifier(random_state=42),
}
model_results = []
for name, model in models.items():
    model.fit(X_train, y_train)
    model_results.append({
        "model": name,
        "train_acc": accuracy_score(y_train, model.predict(X_train)),
        "test_acc": accuracy_score(y_test, model.predict(X_test)),
    })

print(pd.DataFrame(model_results))
```

### Problem 6

Boosting adds trees in sequence, each one fit to what the current model still gets wrong. With a large learning rate every tree moves the model by nearly its full correction, so the early trees chase noise in the training labels and the later ones cannot undo it. With too many trees the training error keeps falling after the real pattern is used up, and from then on each new tree fits noise, so the validation error turns up while the training error still falls. The usual remedy is a small learning rate, with the number of trees chosen where the validation error is lowest.
